```python
class Employee:
    def addEmpDetails(self, E_ID, E_Name, E_Salary, E_Company):
        self.id = E_ID  # (E_ID is a local variables)
        self.name = E_Name
        self.salary = E_Salary
        self.company = E_Company
        #self.company is an instance variable

    def showEmpDetails(self):
        print("Employee ID: ", self.id)
        print("Employee Name: ", self.name)
        print("Employee Salary: ", self.salary)
        print("Employee Company: ", self.company)


e1 = Employee()
e1.addEmpDetails(111, "Raj", 12000.00, "HCL")
e1.showEmpDetails()
print("=" * 25)

e2 = Employee()
e2.addEmpDetails(222, "Ansh Chaudhary", 22000.00, "HCL")
e2.showEmpDetails()
print("=" * 25)
```

```python
class Employee:
    def addEmpDetails(self, E_ID, E_Name, E_Salary, E_Company):    #this keyword
        self.id = E_ID  # (E_ID is a local variables)
        self.name = E_Name
        self.salary = E_Salary
        self.company = E_Company
        # instance variable

    def showEmpDetails(self):
        print("Employee ID: ", self.id)
        print("Employee Name: ", self.name)
        print("Employee Salary: ", self.salary)
        print("Employee Company: ", self.company)


e1 = Employee()
e1.addEmpDetails(111, "Raj", 12000.00, "HCL")   #____new___() __init___()
e1.showEmpDetails()
print("=" * 25)

e2 = Employee()
e2.addEmpDetails(222, "Ansh Chaudhary", 22000.00, "HCL")
e2.showEmpDetails()
print("=" * 25)

  #Constructor are used to initilizde the value for the current objects
  #Employee el=new Employee()； in java programming
```
