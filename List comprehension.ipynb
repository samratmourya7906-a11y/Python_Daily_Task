```python
# without list com....
data = [1,2,3,4,5]
newData=[]
for i in data:
    newData.append(i**2)
print(data)
print(newData)


 # line of code   
```

```python
# with list comprehension
data = [1,2,3,4,5]
newData=[i**2 for i in data]
print(newData)
```

```python
data = [23,12,9,7,26,80,44]
newData=[i for i in data if i%2==0]
print(data)
print(newData)
```

```python
marks=[60,80,89,65,23,87,12,40]
result=["pass" if i>=60 else "fail" for i in marks]
#print(marks)
print(result)
```

```python
names= ["Priya", "Rahul", "Pawan", "John","Suman"]
result=[n.upper() for n in names]

print(result)
```

```python
names= ["  Priya  ", "  Rahul  ", "Pawan", "  John","Suman  "]
result=[n.strip() for n in names]
print(result)
```

```python
names= ["  Priya  ", "  Rahul  ", "Pawan", "  John","Suman  "]
result=[n.lower() for n in names]
print(result)
```
