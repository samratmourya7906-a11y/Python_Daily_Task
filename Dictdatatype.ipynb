```python
# duplicate key not allowed
data = {"name": "mohit", "age":23,"college":"invertis", "name":"amit"}
print(data)
```

```python
#valid keys
data = {1:"amit", "age":23,2.3: "version"}
print(data)
```

```python
#Invalid keys
data = {[1,2]: "Aman"}
print(data)
```

```python
#empty dict
data ={}
print(data)
print(type(data))
data =dict()
print(data)
```

```python
#accessing values
data = {"name": "mohit", "age":23,"college":"invertis"}
#using key
print(data["name"])
#print(data["version"])
# using get method
print(data.get("name"))

```

```python
# adding new element 
data = {"name": "mohit", "age":23,"college":"invertis"}
data["course"]="python programming"
print(data)
```

```python
#updating values
data = {"name": "mohit", "age":23,"college":"invertis"}
data.update({"name":"amit", "age":24})
print(data)
```

```python
#membership 
data = {"name": "mohit", "age":23,"college":"invertis"}
print("age" in data)
print("course" in data)
```

```python
#membership 
data = {"name": "mohit", "age":23,"college":"invertis"}
print("age" not in data)
print("course" not in data)
```

```python
#copy()
data = {"name": "mohit", "age":23,"college":"invertis"}
newData=data.copy()
print(data)
print(newData)
```

```python
#pop()
data = {"name": "mohit", "age":23,"college":"invertis"}
data.pop("age")
print(data)
```
