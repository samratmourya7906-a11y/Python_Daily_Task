```python
s = {45,54,78,98,25,65}
count = 0
total = 0
for i in s:
    count = count + 1
    total = total + i
print("length =", count)
print("s =", count)
```

```python
s = {45,54,78,98,25,65}
maximum = 25
minimum = 25
for i in s:
    if i>maximum:
        maximum = i
    if i<minimum:
        minimum = i
print("maximum =", maximum)
print("minimum =", minimum)
```

```python
s = {45,54,78,98,25,65}
a = list(s)
i = 0
total = 0
while i<len(a):
    total = total + a[i]
    i = i+1
print("sum =", total)
```
