```python
balance=3000  #global variable
def show():   #function without parameter
    global balance
    deposit=1000
    print("deposit balance:", deposit)
    print("initial balance was:", globals() ["balance"])  #one time use
    
show()
print("current balance after deposit:", balance)
```

```python
#Exercise.2
def calculate_discount(bill_amount, discount=None):
    if bill_amount <= 0:
        print("Invalid Bill Amount")
        return

    
    if bill_amount < 1000:
        slab_discount = 5
    elif bill_amount < 5000:
        slab_discount = 10
    elif bill_amount < 10000:
        slab_discount = 15
    else:
        slab_discount = 20

    
    if discount is not None:
        if discount < 0 or discount > 100:
            print("Invalid Discount Percentage")
            return
        discount_percentage = discount
    else:
        discount_percentage = slab_discount

    discount_amount = bill_amount * discount_percentage / 100
    final_amount = bill_amount - discount_amount

    print("Original Bill:", bill_amount)
    print("Discount Percentage:", discount_percentage, "%")
    print("Discount Amount:", discount_amount)
    print("Final Payable Amount:", final_amount)



calculate_discount(800)
calculate_discount(3000)
calculate_discount(7000)
calculate_discount(12000)
calculate_discount(3000, 25)
```

```python
#Excercise.6
def calculate_average():
    try:
        user_input = input("Enter marks separated by spaces: ").strip()

        if not user_input:
            print("Error: Marks cannot be empty.")
            return

        marks = [float(value) for value in user_input.split()]

        
        for mark in marks:
            if mark < 0 or mark > 100:
                print("Error: Marks must be between 0 and 100.")
                return

        average = sum(marks) / len(marks)

        print("Marks:", marks)
        print("Total Marks:", sum(marks))
        print("Average Marks:", average)

    except ValueError:
        print("Error: Please enter valid numeric marks.")



calculate_average()
```

```python
#Student records
students = [
     {"name": "Aman", "marks": [80,75,90]},
     {"name": "Riya", "marks":[35,40,30]},
     {"name": "Rahul", "marks":[90, 85,95]},
     {"name": "Neha","marks":[60, 70, 65]}

]     
def generate_result(student_record):
    for student in student_record:
        name=student  ['name']    #Aman
        marks=student['marks']   #[80,75,90]
        
        avg=sum(marks) /len(marks) #
        if avg>=40:
            status="Pass"
        else:
            status="Fail"
        yield {
           "name":name, 
           "average":avg, 
           "status": status
        }
ex=generate_result(students)
print(next(ex))
print(next(ex))
print(next(ex))
print(next(ex))
```

```python
#Student records
students = [
     {"name": "Aman", "marks": [80,75,90]},
     {"name": "Riya", "marks":[35,40,30]},
     {"name": "Rahul", "marks":[90, 85,95]},
     {"name": "Neha","marks":[60, 70, 65]}

]     
def generate_result(student_record):
    for student in student_record:
        name=student  ['name']    #Aman
        marks=student['marks']   #[80,75,90]
        
        avg=sum(marks) /len(marks) #
        if avg>=40:
            status="Pass"
        else:
            status="Fail"
        yield {
           "name":name, 
           "average":avg, 
           "status": status
        }
for result in generate_result(students):
    print(result['name'])
    print(result['average'])
    print(result['status'])
    print("_"*25)
```
