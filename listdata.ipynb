```python
# method of list
studentNames = ["soniya","amit","raj","ravi","sumit","love" ]
#len()
print(len(studentNames))

#append()
studentNames.append("raj")
print(studentNames)
studentNames.append(["amit","ravi"]) 
print(studentNames)

#insert()
studentNames.insert(0,"rohan")
print(studentNames)
```

```python
# method of list
studentNames = ["soniya","amit","raj","ravi","sumit","love" ]
studentNames.extend(["aman", "rohit"])
print(studentNames)

#remove()
studentNames.remove("raj")
print(studentNames)

studentNames.append("ravi")
studentNames.pop(1)
```
