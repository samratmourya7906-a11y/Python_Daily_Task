```python
# Q.1 write a python program to count a particular character in a given string using while loop.
data="python programming"
strcount=0
i=0
while(i<len(data)):
    if data[i]=='p':
        strcount+=1
    i+=1
print(strcount)
```

```python
# Q.2 write a python program to remove all vowels of any string using for loop 
data="python programming"
result=""
for i in data:
    if i not in "aeiouAEIOU":
        result = result+i
print(result)

```

```python
#Q.3 remove @ from gmail 
data="invertis@gmail.com"
result=""
for i in data:
    if i not in "@":
         result+=i
print(result)
```

```python
#using for loop
data="invertis@gmail.com"
result=""
for i in data:
    if i != "@":
         result+=i
print(result)
```

```python
# using for while loop
data="invertis@gmail.com"
result=""
i=0
while i < len(data):

    if data[i] != "@":
         result+=data[i]
    i+=1
print(result)
```
