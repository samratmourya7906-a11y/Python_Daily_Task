```python
# write a python program atm withrawal
CORRECT_PIN = 1234
account_balance = 10000


entered_pin = int(input("Enter your 4-digit PIN: "))

if entered_pin == CORRECT_PIN:
  
  withdrawal_amount = float(input("Enter amount to withdraw: "))

  
  if withdrawal_amount % 100 == 0:
    
    if withdrawal_amount <= account_balance:
      account_balance -= withdrawal_amount
      print(f"Successfully withdrawn: {withdrawal_amount}")
      print(f"Remaining Balance: {account_balance}")
    else:
      print("Insufficient Balance")
  else:
    print("Error: Amount must be a multiple of 100.")
else:
  
  print("Invalid PIN")
```

```python

sub1 = float(input("Enter marks for Subject 1: "))
sub2 = float(input("Enter marks for Subject 2: "))
sub3 = float(input("Enter marks for Subject 3: "))


if sub1 >= 40 and sub2 >= 40 and sub3 >= 40:
  
  total_marks = sub1 + sub2 + sub3
  percentage = (total_marks / 300) * 100
  print(f"Percentage: {percentage:.2f}%")

  
  if percentage >= 75:
    print("Distinction")
  elif percentage >= 60:
    print("First Division")
  elif percentage >= 50:
    print("Second Division")
  else:
    print("Pass")
else:
  
  print("Fail")
```

```python

cart_amount = float(input("Enter cart amount (₹): "))
is_premium = (
    input("Is the customer a premium member? (yes/no): ").strip().lower()
)


if cart_amount >= 5000:
  if is_premium == "yes":
    discount_percentage = 20  
  else:
    discount_percentage = 10  
else:
  if is_premium == "yes":
    discount_percentage = 5  
  else:
    discount_percentage = 0  


final_amount = cart_amount - (cart_amount * (discount_percentage / 100))

print(f"Discount: {discount_percentage}%")
print(f"Final amount: ₹{final_amount:.2f}")
```

```python
#electricity bill
units = int(input("Enter units consumed: "))
bill = 0


if units <= 100:
  bill = units * 5  
elif units <= 200:
  bill = (100 * 5) + ((units - 100) * 7)  
elif units <= 300:
  bill = (
      (100 * 5) + (100 * 7) + ((units - 200) * 10)
  )  
else:
  bill = (
      (100 * 5) + (100 * 7) + (100 * 10) + ((units - 300) * 15)
  )  


if bill > 3000:
  bill += bill * 0.10

print(f"Total Electricity Bill: ₹{bill:.2f}")
```

```python
#employee bonus
experience = int(input("Enter experience (in years): "))

if experience >= 1:  
  salary = float(input("Enter salary (₹): "))
  performance = float(input("Enter performance score: "))

  if salary >= 50000:
    if performance >= 90:
      bonus_percentage = 20  
    elif performance >= 75:
      bonus_percentage = 10  
    else:
      bonus_percentage = 5  
  else:  
    if performance >= 90:
      bonus_percentage = 15  
    else:
      bonus_percentage = 5  

  bonus_amount = salary * (bonus_percentage / 100)
  print(f"Bonus Percentage: {bonus_percentage}%")
  print(f"Bonus Amount: ₹{bonus_amount:.2f}")
else:
  print("No bonus: Experience must be at least 1 year.")
```
